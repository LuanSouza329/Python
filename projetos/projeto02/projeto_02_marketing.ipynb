# Análise de Clientes, Consumo e Campanhas

## Objetivo

Este projeto tem como objetivo analisar o comportamento dos clientes a partir de dados demográficos, padrões de consumo, canais de compra e respostas às campanhas de marketing.

A análise será realizada utilizando Python, Pandas e Matplotlib, mantendo uma estrutura semelhante ao Projeto 01.

### Principais perguntas

- Qual é a taxa geral de conversão dos clientes?
- Qual é o gasto total e o gasto médio por cliente?
- Quais categorias concentram maior participação nos gastos?
- Quais canais de compra são mais utilizados?
- Qual campanha apresentou melhor taxa de resposta?
- Existem segmentos de clientes com maior potencial de conversão?
- Quais oportunidades de marketing podem ser identificadas?

# 1. Bibliotecas

```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
```

# 2. Carga dos dados

Os dados utilizados neste projeto estão armazenados em um arquivo CSV separado por ponto e vírgula.

A primeira etapa consiste em carregar o conjunto de dados utilizando o Pandas.

```python
DATA_PATH = Path("dados_marketing.csv")

df = pd.read_csv(
    DATA_PATH,
    sep=";",
    parse_dates=["Data Cadastro"],
    dayfirst=True
)

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

df.head()
```

# 3. Preparação e criação das métricas

Antes da análise exploratória, serão criadas métricas iniciais para apoiar a definição dos KPIs do projeto.

```python
colunas_gastos = [
    "Gasto com Eletronicos",
    "Gasto com Brinquedos",
    "Gasto com Moveis",
    "Gasto com Utilidades",
    "Gasto com Alimentos",
    "Gasto com Vestuario",
]

colunas_compras = [
    "Numero de Compras com Desconto",
    "Numero de Compras na Web",
    "Numero de Compras via Catalogo",
    "Numero de Compras na Loja",
]

colunas_campanhas = [
    "Compra na Campanha 1",
    "Compra na Campanha 2",
    "Compra na Campanha 3",
    "Compra na Campanha 4",
    "Compra na Campanha 5",
]

df["gasto_total"] = df[colunas_gastos].sum(axis=1)
df["total_compras"] = df[colunas_compras].sum(axis=1)
df["total_filhos"] = df["Filhos em Casa"] + df["Adolescentes em Casa"]
df["respostas_campanhas"] = df[colunas_campanhas].sum(axis=1)

ano_referencia = df["Data Cadastro"].dt.year.max()
df["idade_aproximada"] = ano_referencia - df["Ano Nascimento"]

df.head()
```

# 4. KPIs

Indicadores chave de desempenho definidos para avaliar o comportamento dos clientes, o volume de consumo, a conversão e a performance das campanhas de marketing.

```python
total_clientes = df["ID"].nunique()

clientes_compraram = df["Comprou"].sum()

taxa_conversao = (
    clientes_compraram /
    total_clientes *
    100
).round(2)

gasto_total = df["gasto_total"].sum()

gasto_medio_cliente = (
    gasto_total /
    total_clientes
).round(2)

total_compras = df["total_compras"].sum()

ticket_medio_estimado = (
    gasto_total /
    total_compras
).round(2)

salario_medio = df["Salario Anual"].mean().round(2)

recencia_media = df["Dias Desde Ultima Compra"].mean().round(2)

categoria_maior_gasto = df[colunas_gastos].sum().idxmax().replace("Gasto com ", "")

canal_mais_usado = df[colunas_compras].sum().idxmax().replace("Numero de Compras ", "")

campanha_maior_resposta = df[colunas_campanhas].sum().idxmax().replace("Compra na ", "")
```

# 5. DataFrame KPIs

```python
df_kpis = pd.DataFrame({
    "total_clientes": [total_clientes],
    "clientes_compraram": [clientes_compraram],
    "taxa_conversao": [taxa_conversao],
    "gasto_total": [gasto_total],
    "gasto_medio_cliente": [gasto_medio_cliente],
    "total_compras": [total_compras],
    "ticket_medio_estimado": [ticket_medio_estimado],
    "salario_medio": [salario_medio],
    "recencia_media": [recencia_media],
    "categoria_maior_gasto": [categoria_maior_gasto],
    "canal_mais_usado": [canal_mais_usado],
    "campanha_maior_resposta": [campanha_maior_resposta],
})

df_kpis
```
